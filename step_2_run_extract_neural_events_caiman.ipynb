```python
#Note: if the multiprocessing / parallel processing doesn't work and you can't fix easily
#you can just loop through the input files in a for loop and run run_notebook on each file
#it will generally take slightly longer to run but will still work ok.
import os
import re #for regular expressions
import nbformat 
from nbconvert.preprocessors import ExecutePreprocessor
import sys
import pandas as pd

# Define global paths
import_dir= "/Users/cybellesmith/Box Sync/all/post_doc2/code/kiera_data_analysis/processed_videos"
event_dir = "/Users/cybellesmith/Box Sync/all/post_doc2/code/kiera_data_analysis/event_detection_output"
template_analysis_dir = "/Users/cybellesmith/Box Sync/all/post_doc2/code/kiera_data_analysis/code"
template_analysis_file = 'step_2_extract_neural_events_caiman_template.ipynb'
jupyter_notebook_out_path = "/Users/cybellesmith/Box Sync/all/post_doc2/code/kiera_data_analysis/code/jupyter_notebooks"

#make the directory for storing the jupyter notebooks you generate if it doesn't exist yet
os.makedirs(jupyter_notebook_out_path, exist_ok=True)

def run_notebook(image_file):

    image_id = re.sub(r"\_mc_dff_resid.tif$", "_caiman_analysis", image_file)
    
    #read in the template jupyter notebook
    os.chdir(template_analysis_dir)      
    with open(template_analysis_file, encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)

    #define input parameters to add to first cell of the template notebook
    input_parameters = {'video_filename': image_file, 'video_dir': import_dir, 'event_dir': event_dir}

    print(f"Input parameters: {image_file}")

    #finds the first cell in the template file capable of being written to...
    parameters_cell_index = -1
    for i, cell in enumerate(nb['cells']):
        if 'tags' in cell['metadata'] and 'parameters' in cell['metadata']['tags']:
            parameters_cell_index = i
            break

    if parameters_cell_index == -1:
        parameters_cell_index = 0
        nb['cells'].insert(0, nbformat.v4.new_code_cell(''))

    #convert dictionary of parameters specific to the image (its filename etc.) into code
    #that you will paste into the parameters cell (generally the first cell) of the new jupyter notebook
    param_code = '\n'.join([f'{k} = {repr(v)}' for k, v in input_parameters.items()])
    
    #write code to the parameters cell in the new jupyter notebook
    nb['cells'][parameters_cell_index]['source'] = param_code

    #adds metadata to the cell
    if 'tags' not in nb['cells'][parameters_cell_index]['metadata']:
        nb['cells'][parameters_cell_index]['metadata']['tags'] = []
    if 'parameters' not in nb['cells'][parameters_cell_index]['metadata']['tags']:
        nb['cells'][parameters_cell_index]['metadata']['tags'].append('parameters')

    #run the newly created jupyter notebook (generating and saving figures, spreadsheets, etc.)
    kernel_name = nb.metadata.get("kernelspec", {}).get("name", "python3")
    ep = ExecutePreprocessor(timeout=600, kernel_name=kernel_name)

    try:
        ep.preprocess(nb, {'metadata': {'path': template_analysis_dir}})
    except Exception as e:
        print(f"Error executing notebook {image_file}: {e}", file=sys.stderr)
        return

    #save the newly created jupyter notebook with all internal plots specific to the input image in the notebook
    output_path = os.path.join(jupyter_notebook_out_path, f"{image_id}.ipynb")
    with open(output_path, 'w', encoding='utf-8') as f:
        nbformat.write(nb, f)

    print(f"Executed notebook saved to {output_path}")

video_files_mc = [
'DIV27_CTRL-CTRL_W5_optostim_mc.tif',
 'DIV27_CTRL-CTRL_W4_optostim_mc.tif',
 'DIV27_CTRL-CTRL_W3_optostim_mc.tif',
 'DIV27_HS1-DA1M_W7_optostim_mc.tif',
 'DIV27_HS1-DA1M_W8_optostim_mc.tif',
 'DIV27_HS1-DA1M_W11_optostim_mc.tif',
 'DIV27_HS1-DA1M_W10_optostim_mc.tif',
 'DIV27_HS1-DA1M_W9_optostim_mc.tif'  
]

for file in video_files_mc:
    video_file = re.sub(r"\.tif$", "_dff_resid.tif", file)
    print(video_file)
    run_notebook(video_file)
```
